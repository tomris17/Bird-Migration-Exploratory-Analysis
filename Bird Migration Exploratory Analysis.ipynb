# Bird Migration Exploratory Analysis

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("bird_migration_data.csv")
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    x="Species",
    y="Flight_Distance_km",
    data=df,
    palette="muted",
    estimator="mean",
    hue="Species",
    legend=False,
)
plt.title(
    "Average Flight Distance by Bird Species", fontsize=14, fontweight="bold"
)
plt.xlabel("Species", fontsize=12)
plt.ylabel("Average Flight Distance (km)", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(
    x="Migration_Reason",
    data=df,
    palette="Set2",
    order=df["Migration_Reason"].value_counts().index,
)
plt.title("Distribution of Migration Reasons", fontsize=14, fontweight="bold")
plt.xlabel("Migration Reason", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x="Temperature_C",
    y="Flight_Distance_km",
    hue="Weather_Condition",
    data=df.sample(500),
    alpha=0.7,
    palette="tab10",
)
plt.title(
    "Temperature vs Flight Distance by Weather Condition",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Temperature (°C)", fontsize=12)
plt.ylabel("Flight Distance (km)", fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```
